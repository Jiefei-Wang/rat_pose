```python
import pandas as pd

# Get available models
available = set(available_models())

# List to store valid combinations
combinations = []

# Manually extracted combinations based on the model paths and available models
data = [
    # Full body models
    ("full_human", "resnet_101"),
    ("full_cat", "resnet_50"),
    ("full_dog", "resnet_50"),
    ("primate_face", "resnet_50"),
    ("mouse_pupil_vclose", "resnet_50"),
    ("horse_sideview", "resnet_50"),
    ("full_macaque", "resnet_50"),
    
    # SuperAnimal TopViewMouse
    ("superanimal_topviewmouse_dlcrnet", "resnet_50"),
    ("superanimal_topviewmouse", "resnet_50"),
    ("superanimal_topviewmouse", "hrnet_w32"),
    ("superanimal_topviewmouse", "resnet_50"),
    
    # SuperAnimal Quadruped
    ("superanimal_quadruped_dlcrnet", "resnet_50"),
    ("superanimal_quadruped", "resnet_50"),
    ("superanimal_quadruped", "hrnet_w32"),
    ("superanimal_quadruped", "resnet_50"),
    ("superanimal_quadruped", "rtmpose_s"),
    ("superanimal_quadruped", "rtmpose_m"),
    ("superanimal_quadruped", "rtmpose_x"),
    ("superanimal_quadruped", "hrnet_w48"),
    ("superanimal_quadruped", "resnet_101"),
    
    # SuperAnimal Bird
    ("superanimal_bird", "resnet_50"),
    
    # SuperAnimal HumanBody
    ("superanimal_humanbody", "rtmpose_x"),
]

# CRITICAL: Verify that "{animal_name}_{model_name}" exists in model_names
# because download_super_animal_snapshot() creates the combined key
for animal_name, model_name in data:
    combined_key = f"{animal_name}_{model_name}"
    
    # Check if the combined key exists AND model is in available list
    if combined_key in model_names and model_name in available:
        combinations.append({
            "animal_name": animal_name, 
            "model_name": model_name,
            "combined_key": combined_key
        })

# Create DataFrame
df = pd.DataFrame(combinations)
df.drop_duplicates(inplace=True)
df
```
